```python
import spacy
from collections import Counter
import re
import random

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Clean tweet text
def clean_tweet(text):
    text = re.sub(r"http\S+|@\S+|#[A-Za-z0-9_]+", "", text)  # Remove URLs, mentions, hashtags
    text = re.sub(r"[^A-Za-z0-9\s]", "", text)  # Remove special characters
    return text.strip()

# Simulated fallback because snscrape is deprecated and broken in Python 3.13
def mine_tweets_simulated(keyword, since, until, max_tweets=100):
    print(f"[Simulation] Querying tweets for '{keyword}' from {since} to {until}...")

    # Template templates to generate realistic NLP processing targets
    sample_templates = [
        "Just heard Google is launching a new AI model in San Francisco! #tech",
        "I love visiting London during the spring. The weather is beautiful! @traveler",
        "Elon Musk announced new updates for Tesla. Is this the future of transportation? #Tesla",
        "Apple is planning to host its next big event in California next month.",
        "Microsoft and OpenAI are collaborating on massive new infrastructure plans in the United States.",
        "Had a great dinner in Tokyo, Japan. The food here is absolutely incredible!",
        "Breaking: United Nations addresses global climate initiatives at the latest summit in Geneva.",
        "Can't wait for the next SpaceX launch next Tuesday! @SpaceX #space"
    ]

    tweets = []
    for _ in range(min(max_tweets, 50)):
        tweets.append(random.choice(sample_templates))
    return tweets

# Extract trends and named entities
def analyze_tweets(tweets):
    all_text = " ".join(clean_tweet(t) for t in tweets)
    words = [word.lower() for word in all_text.split() if len(word) > 3]
    word_freq = Counter(words).most_common(10)

    # Named entity recognition
    entity_counter = Counter()
    for text in tweets:
        doc = nlp(clean_tweet(text))
        for ent in doc.ents:
            if ent.label_ in ['PERSON', 'ORG', 'GPE']:
                entity_counter[ent.text] += 1

    return word_freq, entity_counter.most_common(10)

# Example Usage
if __name__ == "__main__":
    keyword = input("Enter keyword or hashtag (optional): ").strip() or "AI"
    since = input("Enter start date (YYYY-MM-DD): ").strip() or "2023-01-01"
    until = input("Enter end date (YYYY-MM-DD): ").strip() or "2023-01-08"

    # Using simulation since snscrape is deprecated
    tweets = mine_tweets_simulated(keyword, since, until, max_tweets=200)
    print(f"\nCollected {len(tweets)} tweets.")

    trends, entities = analyze_tweets(tweets)

    print("\n--- Trending Words ---")
    for word, freq in trends:
        print(f"{word}: {freq}")

    print("\n--- Named Entities (People, Orgs, Places) ---")
    for ent, freq in entities:
        print(f"{ent}: {freq}")
```
